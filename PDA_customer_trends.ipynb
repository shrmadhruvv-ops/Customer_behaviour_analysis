```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
df.head(50)
```

```python
# Create a column age group
labels = ['Adult','Young Adult','Middle Aged','Senior'] 
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels) # qcut assigns the labels into equal distributions in a range
```

```python
df['frequency_of_purchases'].unique().tolist()
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly':14,
 'Weekly':7,
 'Annually':365,
 'Quarterly':60,
 'Bi-Weekly':14,
 'Monthly':30,
 'Every 3 Months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df.head(50)
```

```python
(df['promo_code_used']==df['discount_applied']).all() 
```

```python
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
df
```

```python
df = df.drop('promo_code_used', axis = 1)
df.head(10)
```
